# Inspect and spawn independent request contexts

Five learned manager adapters serve200 registered request slots per model family. The weights are shared and immutable; each row starts with an empty history.

```python
from pathlib import Path
import sys, json
YOUR_PATHWAY = "YOUR PATHWAY"
if YOUR_PATHWAY == "YOUR PATHWAY":
    ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/"Models/adapter_manifest.json").is_file()), None)
    if ROOT is None:
        raise ValueError("Set YOUR_PATHWAY to the repository clone directory.")
else:
    ROOT = Path(YOUR_PATHWAY).expanduser().resolve()
sys.path[:0] = [str(ROOT/"Experiment Codes/src"), str(ROOT/"Parallel Agent Setup/src")]
RUN_INFERENCE = False
RUN_TRAINING = False
```

```python
from workbench_protocol import build_cases, spawn_instance
for family in ["Mistral", "Qwen"]:
    records = json.loads((ROOT/"Models"/family/"manager_instance_registry.json").read_text())
    cases = build_cases(1, ROOT/"Experiment Codes"/family)
    assert len(records) == len(cases) == 200
    assert len({c["agent_instance"]["instance_id"] for c in cases}) == 200
    assert all(not c["agent_instance"]["history"] for c in cases)
    print(family, "instances:", len(cases), "trained adapters:", len({c["manager"] for c in cases}))
```
